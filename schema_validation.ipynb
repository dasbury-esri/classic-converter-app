```python
import json
from storymap_json_schema import validate_storymap_json

# Path to your draft_*.json file
with open("test_data/storymaps/draft_1761795177964.json", "r", encoding="utf-8") as f:
    storymap_json = json.load(f)

# Validate the whole storymap JSON
errors = validate_storymap_json(storymap_json)

if errors:
    print("Validation errors found:")
    for err in errors:
        print("-", err)
else:
    print("No validation errors found. StoryMap JSON is valid.")
```

```python
from storymap_json_schema import validate_node_against_schema

for node_id, node in storymap_json["nodes"].items():
    node_type = node.get("type")
    node_errors = validate_node_against_schema(node, node_type)
    if node_errors:
        print(f"Errors in node {node_id} (type: {node_type}):")
        for err in node_errors:
            print("-", err)
    else:
        print(f"Node {node_id} is valid. (type: {node_type})")
```
